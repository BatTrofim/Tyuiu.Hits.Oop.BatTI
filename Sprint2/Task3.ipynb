```
var bachelor = new BachelorStudent("Анна", 20, 3, "ИТ", 4.7,
    "Информационные технологии", "Веб-приложение");
bachelor.GetInfo();
bachelor.Study();
bachelor.TakeExam();
bachelor.ShowFaculty();
bachelor.DefendCoursework();
Console.WriteLine($"Стипендия: {bachelor.CalculateScholarship()} руб.\n");

var master = new MasterStudent("Олег", 23, 1, "ИТ", 4.9,
    "Профессор Петров И.С.", "Машинное обучение");
master.GetInfo();
master.Study();
master.TakeExam();
master.ConductResearch();
Console.WriteLine();

var phd = new PhDStudent("Елена", 25, 2, "ИТ", 5.0,
    "Профессор Петров И.С.", "Нейронные сети",
    "Оптимизация нейронных сетей", 3);
phd.GetInfo();
phd.Study();
phd.TakeExam();
phd.ConductResearch();
phd.Publish();
Console.WriteLine();

var bachelor2 = new BachelorStudent("Игорь", 21, 3, "ИТ", 4.3,
    "Кибербезопасность", "Анализ уязвимостей");
Interaction.Interact(bachelor, bachelor2);
Interaction.Interact(master, phd);
Interaction.ResearchTogether(master, phd);
public interface IResearcher
{
    string ResearchArea { get; set; }
    void ConductResearch();
}

public interface IPublisher
{
    int PublicationsCount { get; set; }
    void Publish();
}

public class Student
{
    public string Name { get; set; }
    public int Age { get; set; }
    public int Course { get; set; }

    public string Faculty { get; set; }
    public double AverageGrade { get; set; }

    public Student(string name, int age, int course, string faculty, double averageGrade)
    {
        Name = name;
        Age = age;
        Course = course;
        Faculty = faculty;
        AverageGrade = averageGrade;
    }

    public virtual void GetInfo() =>
        Console.WriteLine($"Имя: {Name}, Возраст: {Age}, Курс: {Course}");

    public virtual void Study() =>
        Console.WriteLine($"{Name} сейчас учится.");

    public virtual void TakeExam() =>
        Console.WriteLine($"{Name} сдает экзамен.");

    public void ShowFaculty() =>
        Console.WriteLine($"{Name} учится на факультете \"{Faculty}\".");

    public double CalculateScholarship() =>
        AverageGrade >= 4.5 ? 3500 : 2000;
}

public class BachelorStudent : Student
{
    public string Specialty { get; set; }
    public string ThesisTopic { get; set; }

    public BachelorStudent(string name, int age, int course, string faculty,
                           double averageGrade, string specialty, string thesisTopic)
        : base(name, age, course, faculty, averageGrade)
    {
        Specialty = specialty;
        ThesisTopic = thesisTopic;
    }

    public override void Study() =>
        Console.WriteLine($"{Name} изучает предметы по специальности \"{Specialty}\".");

    public void DefendCoursework() =>
        Console.WriteLine($"{Name} защищает курсовую: \"{ThesisTopic}\".");
}

public class MasterStudent : Student, IResearcher
{
    public string ScientificAdvisor { get; set; }
    public string ResearchArea { get; set; }

    public MasterStudent(string name, int age, int course, string faculty,
                         double averageGrade, string advisor, string researchArea)
        : base(name, age, course, faculty, averageGrade)
    {
        ScientificAdvisor = advisor;
        ResearchArea = researchArea;
    }

    public override void TakeExam() =>
        Console.WriteLine($"{Name} сдает экзамен. Научный руководитель: {ScientificAdvisor}.");

    public void ConductResearch() =>
        Console.WriteLine($"{Name} исследует область \"{ResearchArea}\".");
}

public class PhDStudent : MasterStudent, IPublisher
{
    public string DissertationTopic { get; set; }
    public int PublicationsCount { get; set; }

    public PhDStudent(string name, int age, int course, string faculty,
                      double averageGrade, string advisor, string researchArea,
                      string dissertationTopic, int publicationsCount)
        : base(name, age, course, faculty, averageGrade, advisor, researchArea)
    {
        DissertationTopic = dissertationTopic;
        PublicationsCount = publicationsCount;
    }

    public override void Study() =>
        Console.WriteLine($"{Name} исследует тему: \"{DissertationTopic}\".");

    public void Publish()
    {
        PublicationsCount++;
        Console.WriteLine($"{Name} опубликовал статью. Всего: {PublicationsCount}.");
    }
}

public static class Interaction
{
    public static void Interact(Student s1, Student s2)
    {
        if (s1 is BachelorStudent b1 && s2 is BachelorStudent b2)
            Console.WriteLine($"{b1.Name} и {b2.Name} обсуждают курсовые: \"{b1.ThesisTopic}\" и \"{b2.ThesisTopic}\".");
        else
            Console.WriteLine($"{s1.Name} ({s1.GetType().Name}) взаимодействует с {s2.Name} ({s2.GetType().Name}).");
    }

    public static void ResearchTogether(IResearcher r1, IResearcher r2) =>
        Console.WriteLine($"Совместное исследование: \"{r1.ResearchArea}\" + \"{r2.ResearchArea}\".");
}
```
