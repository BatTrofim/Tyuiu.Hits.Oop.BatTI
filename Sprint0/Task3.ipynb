### Вариант задания 2



<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Создать базовый класс Project в C#, который будет представлять проекты в рамках
организации. На основе этого класса разработать 2-3 производных класса,
демонстрирующих принципы наследования и полиморфизма. В каждом из классов
должны быть реализованы новые атрибуты и методы, а также переопределены
некоторые методы базового класса для демонстрации полиморфизма.

<h2 style="color:DodgerBlue">Требования к базовому классу Project:</h2>

• Атрибуты: Имя (Name), Возраст (Age), Курс (Course).

• Методы:
• GetInfo(): метод для получения информации о студенте в виде строки.
• Study(): метод для вывода сообщения о том, что студент учится.
• TakeExam(): метод для вывода сообщения о сдаче экзамена.

<h2 style="color:DodgerBlue">Требования к производным классам:</h2>

1. Студент бакалавриата (BachelorStudent): Должен содержать дополнительные
 атрибуты, такие как Специальность (Specialty). Метод Study() должен быть
  переопределен для добавления информации о специальности при изучении
   предмета.

2. Студент магистратуры (MasterStudent): Должен содержать дополнительные
 атрибуты, такие как Научный руководитель (ScientificAdvisor).
  Метод TakeExam() должен быть переопределен для добавления
   информации о научном руководителе при сдаче экзамена.

3. Аспирант (PhDStudent) (если требуется третий класс): Должен содержать
 дополнительные атрибуты, такие как Тема диссертации (DissertationTopic).
  Метод Study() должен быть переопределен для добавления информации о теме диссертации при изучении предмета.



<h2 style="color:DodgerBlue">Реализация:</h2>

----

```
BachelorStudent bachelor = new BachelorStudent();
bachelor.Name = "Анна";
bachelor.Age = 20;
bachelor.Course = 3;
bachelor.Specialty = "Информационные технологии";
bachelor.GetInfo();
bachelor.Study();
bachelor.TakeExam();
Console.WriteLine();

MasterStudent master = new MasterStudent();
master.Name = "Олег";
master.Age = 23;
master.Course = 1;
master.ScientificAdvisor = "Профессор Петров И.С.";
master.GetInfo();
master.Study();
master.TakeExam();
Console.WriteLine();

PhDStudent phd = new PhDStudent();
phd.Name = "Елена";
phd.Age = 25;
phd.Course = 2;
phd.DissertationTopic = "Оптимизация нейронных сетей";
phd.GetInfo();
phd.Study();
phd.TakeExam();
Console.WriteLine();

public class Project
{
    public string Name;
    public int Age;
    public int Course;

    public virtual void GetInfo()
    {
        Console.WriteLine($"Имя: {Name}, Возраст: {Age}, Курс: {Course}");
    }

    public virtual void Study()
    {
        Console.WriteLine($"{Name} сейчас учится.");
    }

    public virtual void TakeExam()
    {
        Console.WriteLine($"{Name} сдает экзамен.");
    }
}

public class BachelorStudent : Project
{
    public string Specialty;

    public override void Study()
    {
        Console.WriteLine($"{Name} изучает предметы по специальности \"{Specialty}\".");
    }
}

public class MasterStudent : Project
{
    public string ScientificAdvisor;

    public override void TakeExam()
    {
        Console.WriteLine($"{Name} сдает экзамен. Присутствует научный руководитель: {ScientificAdvisor}.");
    }
}

public class PhDStudent : Project
{
    public string DissertationTopic;

    public override void Study()
    {
        Console.WriteLine($"{Name} проводит исследование по теме: \"{DissertationTopic}\".");
    }
}
```
