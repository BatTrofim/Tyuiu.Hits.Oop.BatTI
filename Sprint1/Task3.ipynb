```
BachelorStudent bachelor = new BachelorStudent("Анна", 20, 3, "Информационные технологии");
bachelor.GetInfo();
bachelor.Study();
bachelor.TakeExam();
Console.WriteLine();

MasterStudent master = new MasterStudent("Олег", 23, 1, "Профессор Петров И.С.");
master.GetInfo();
master.Study();
master.TakeExam();
Console.WriteLine();

PhDStudent phd = new PhDStudent("Елена", 25, 2, "Оптимизация нейронных сетей");
phd.GetInfo();
phd.Study();
phd.TakeExam();
Console.WriteLine();

Project.Interact(bachelor, master);
Project.Interact(master, phd);
Console.WriteLine();

BachelorStudent anotherBachelor = new BachelorStudent("Игорь", 21, 3, "Кибербезопасность");
Project.Interact(bachelor, anotherBachelor);

public class Project
{
    public string Name { get; set; }
    public int Age { get; set; }
    public int Course { get; set; }

    public Project() { }

    public Project(string name, int age, int course)
    {
        Name = name;
        Age = age;
        Course = course;
    }

    public virtual void GetInfo()
    {
        Console.WriteLine($"Имя: {Name}, Возраст: {Age}, Курс: {Course}");
    }

    public virtual void Study()
    {
        Console.WriteLine($"{Name} сейчас учится.");
    }

    public virtual void TakeExam()
    {
        Console.WriteLine($"{Name} сдает экзамен.");
    }

    public static void Interact(Project p1, Project p2)
    {
        if (p1 == null || p2 == null)
        {
            Console.WriteLine("Неполное взаимодействие: один из объектов равен null.");
            return;
        }

        if (p1 is BachelorStudent b1 && p2 is BachelorStudent b2)
        {
            Console.WriteLine($"{b1.Name} и {b2.Name} обсуждают курсовой проект по специальности \"{b1.Specialty}\" и \"{b2.Specialty}\".");
            return;
        }

        Console.WriteLine($"{p1.Name} ({p1.GetType().Name}) взаимодействует с {p2.Name} ({p2.GetType().Name}).");
    }
}

public class BachelorStudent : Project
{
    public string Specialty { get; set; }

    public BachelorStudent() { }

    public BachelorStudent(string name, int age, int course, string specialty)
    {
        Name = name;
        Age = age;
        Course = course;
        Specialty = specialty;
    }

    public override void Study()
    {
        Console.WriteLine($"{Name} изучает предметы по специальности \"{Specialty}\".");
    }
}

public class MasterStudent : Project
{
    public string ScientificAdvisor { get; set; }

    public MasterStudent() { }

    public MasterStudent(string name, int age, int course, string advisor)
    {
        Name = name;
        Age = age;
        Course = course;
        ScientificAdvisor = advisor;
    }

    public override void TakeExam()
    {
        Console.WriteLine($"{Name} сдает экзамен. Присутствует научный руководитель: {ScientificAdvisor}.");
    }
}

public class PhDStudent : Project
{
    public string DissertationTopic { get; set; }

    public PhDStudent() { }

    public PhDStudent(string name, int age, int course, string topic)
    {
        Name = name;
        Age = age;
        Course = course;
        DissertationTopic = topic;
    }

    public override void Study()
    {
        Console.WriteLine($"{Name} проводит исследование по теме: \"{DissertationTopic}\".");
    }
}
```
